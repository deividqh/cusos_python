```python
import numpy as np

# calculo de pesos:
def calcular_activacion(entradas, pesos, sesgo=0):
    """
    Simula la activación de una neurona.
    Entradas: Vectores(tensor de rango 1)
    pesos: Vectores(misma dimensión que entradas)
    """
    # Producto escalar.
    activacion = np.dot(entradas, pesos) + sesgo
    return activacion


# prueba1
x = np.array([1.5, 2.0, -0.5]) # ejemplo de entradas [ edad, ingresos, historial ]
w = np.array([0.8, 0.1, 0.4]) # importancias de los valores de entrada son los pesos sinapticos
resultado_activacion = calcular_activacion(x, w)
print("Resultado de la activación:", resultado_activacion)




```

```python
import numpy as np
def resolver_sistema_lineal(A, B):
    """
        Resuelve un sistema de ecuaciones lineales Ax = B
    """
    print(f'• A)\n {A}')
    print(f'• B) \n{B}')
    try:
        # Calculo de la inversa de A
        a = np.linalg.inv(A)
        # print(f'\n• inversa \n{inversa}')

        # Calculo del producto matricial
        b = np.dot( A, B )

        # Producto Matricial
        c = A.dot(B)

        # Transponer una matriz
        d = A.T

        return a, b, c , d
    except np.linalg.LinAlgError as e:
        print (f"Error al resolver el sistema: {e}")
        return None , None

"""
"""
A = np.array([[1, 1, 1], [1, 2, 1], [1, 3, 2]])
B = np.array([-1, -2, 3])

a, b, c, d = resolver_sistema_lineal(A, B)
print("\n■ Solucion")
print(f'\n• inversa: \n{a}')
print(f'\n• producto matricial 1:\n{b}')
print(f'\n• producto matricial 2:\n{c}')
print(f'\n• Transpuesta de A:\n{d}')
```
